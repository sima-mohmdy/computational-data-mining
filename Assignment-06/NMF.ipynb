# purity

```python
def purity_score(y_true, y_pred):
    # compute contingency matrix (also called confusion matrix)
    contingency_matrix = metrics.cluster.contingency_matrix(y_true, y_pred)
    # return purity
    return np.sum(np.amax(contingency_matrix, axis=0)) / np.sum(contingency_matrix)
```

# entropy

```python
def entropy_score(y_true, y_pred):
    # compute contingency matrix (also called confusion matrix)
    contingency_matrix = metrics.cluster.contingency_matrix(y_true, y_pred)

    sum_of_each_colum = np.sum(contingency_matrix,axis = 0)

    entropy = 0
    for i in range(contingency_matrix.shape[0]):
      for j in range(contingency_matrix.shape[1]):
        if contingency_matrix[i, j] != 0:
          entropy += (-1) * (contingency_matrix[i, j]) * (np.log(contingency_matrix[i, j] / sum_of_each_colum[j]))

    n = len(y_true)
    m = len(np.unique(y_true))
    entropy *= (1 /(n * np.log(m)))

    return entropy

```

# NMF Clustering

```python
def predict_clusters_with_nmf(X, y_true, n_components):
  nmf_model = NMF(n_components=n_components, init='random', random_state=42)
  # Fitting the NMF model
  W = nmf_model.fit_transform(X)  # Feature matrix
  H = nmf_model.components_                # Coefficient matrix

  y_pred = np.argmax(W, axis=1)

  purity = purity_score(y_true.flatten(), y_pred)
  entropy = entropy_score(y_true.flatten(), y_pred)

  return purity, entropy
```

## K-means Clustering

```python
def predict_clusters_with_kmeans(X, y_true, n_clusters):
  kmeans = KMeans(n_clusters=n_clusters, random_state=0, n_init="auto").fit(X)
  y_pred = kmeans.predict(X)

  purity = purity_score(y_true.flatten(), y_pred)
  entropy = entropy_score(y_true.flatten(), y_pred)

  return purity, entropy
```
